Dataset: Retail Demand Forecasting

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
import pandas as pd

df = pd.read_csv("../data/processed/retail_clean.csv")
```

```python
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
markdown_cols = [
    "MarkDown1",
    "MarkDown2",
    "MarkDown3",
    "MarkDown4",
    "MarkDown5"
]

df[markdown_cols] = df[markdown_cols].fillna(0)
```

```python
df.isnull().sum()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.dtypes
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
df.describe().T
```

```python
df.to_csv("../data/processed/retail_clean.csv", index=False)
```
