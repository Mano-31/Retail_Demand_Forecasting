# Feature Engineering

Project: AI-Powered Retail Demand Forecasting & Inventory Optimization

Objective:

Create meaningful features that improve forecasting accuracy.

```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv("../data/processed/retail_clean.csv")
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Week"] = df["Date"].dt.isocalendar().week.astype(int)
df["Quarter"] = df["Date"].dt.quarter
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.dayofweek
```

```python
df["Weekend"] = (df["DayOfWeek"] >= 5).astype(int)
```

```python
df = df.sort_values(["Store", "Dept", "Date"])

df["Lag_1"] = (
    df.groupby(["Store", "Dept"])["Weekly_Sales"]
      .shift(1)
)
```

```python
df["Lag_2"] = (
    df.groupby(["Store", "Dept"])["Weekly_Sales"]
      .shift(2)
)
```

```python
df["Rolling_Mean_4"] = (
    df.groupby(["Store", "Dept"])["Weekly_Sales"]
      .transform(lambda x: x.rolling(4).mean())
)
```

```python
df["Rolling_STD_4"] = (
    df.groupby(["Store", "Dept"])["Weekly_Sales"]
      .transform(lambda x: x.rolling(4).std())
)
```

```python
df["Total_Markdown"] = (
    df["MarkDown1"] +
    df["MarkDown2"] +
    df["MarkDown3"] +
    df["MarkDown4"] +
    df["MarkDown5"]
)
```

```python
df["Sales_per_Size"] = df["Weekly_Sales"] / df["Size"]
```

```python
lag_cols = [
    "Lag_1",
    "Lag_2",
    "Rolling_Mean_4",
    "Rolling_STD_4"
]

df[lag_cols] = df[lag_cols].fillna(0)
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=["Type"], drop_first=True)

print(df.columns.tolist())

df.to_csv("../data/processed/retail_featured.csv", index=False)

print("Feature engineering completed successfully!")
```
